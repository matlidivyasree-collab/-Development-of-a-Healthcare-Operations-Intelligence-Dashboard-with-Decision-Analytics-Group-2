```python
import pandas as pd
from ydata_profiling import ProfileReport
```

```python
from pathlib import Path

# Resolve CSV path whether the notebook is run from repo root or cleaning/ folder.
data_path = Path("fact_outbreak_clean.csv")
if not data_path.exists():
    data_path = Path("cleaning") / "fact_outbreak_clean.csv"

data = pd.read_csv(data_path)
data.head()
```

```python
print("Helloworld")
```

```python
# Use a bounded sample and lightweight settings to avoid kernel crashes on large datasets.

profile = ProfileReport(
    data,
    title="Outbreak Data Profiling Report",
    minimal=True,
    explorative=False,
    correlations={"auto": {"calculate": False}},
    interactions={"continuous": False},
)
profile.to_notebook_iframe()
```

```python
profile.to_file('Eda_Report_outbreak.html')
```
