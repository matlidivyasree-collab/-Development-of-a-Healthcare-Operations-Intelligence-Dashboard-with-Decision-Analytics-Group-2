```python
## seperate columns with the numerical and categorical 
num=[]
cat=[]
for i in data.columns:
    if data[i].dtype=='int64' or data[i].dtype=='float64':
        num.append(i)
    else:
        cat.append(i)
print('Numerical Columns :',num)
print("Categorical Columns :",cat)
```

```python
# checking the value count of the categorical columns 
for i in cat:
    print(data[i].value_counts())
```

```python
data['new_outbreak_flag'].value_counts()
```

```python
count = 0
for i in range(len(data['controlled_flag'])):
    if data['controlled_flag'][i]=='TRUE':
        count += 1
print(count)
```

```python
# cleaning column : new_outbreak_flag
print(data['new_outbreak_flag'].value_counts())
data['new_outbreak_flag'] = data['new_outbreak_flag'].replace(['True','Y','Yes','YES','yes','1'], 'True',inplace=True)
data['new_outbreak_flag'] = data['new_outbreak_flag'].replace(['False','N','No','NO','0'], 'False',inplace=True)
print("After cleaning")
print(data['new_outbreak_flag'].value_counts())
```

```python
## Cleaning for this column : controlled_flag 

print("Before cleaning ")
print(data['controlled_flag'].value_counts())
data['controlled_flag']=data['controlled_flag'].replace(['True','TRUE','Yes','Y','1',1,'YES','yes',],True,inplace=True)
data['controlled_flag']=data['controlled_flag'].replace(['False','FALSE','no','N','No','NO',0,'0'],False,inplace=True)
print("After cleaning ")
data['controlled_flag'].value_counts()
```

```python
data['alert_level']=data['alert_level'].str.strip()  # Remove extra space in categories 

data['alert_level']=data['alert_level'].replace(['Low','LOW','low',' Low','low' '  Low'],'Low',inplace=True)
data['alert_level']=data['alert_level'].replace(['High',' High','high','HIGH','  High'],'High',inplace=True)
data['alert_level']=data['alert_level'].replace(['Moderate','MODERATE','moderate',' Moderate','  Moderate'],'Moderate',inplace=True)
```
